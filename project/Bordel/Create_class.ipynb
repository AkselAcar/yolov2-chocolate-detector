# Check new target boxes are ok :

```python
import unicodedata
import json
import pandas as pd


def normalize_name(s):
    # NFKD décompose les caractères accentués, on jette tout ce qui n'est pas ASCII
    s = unicodedata.normalize('NFKD', s)
    s = s.encode('ascii', 'ignore').decode('ascii')
    # remplace espaces et underscores par un underscore, et tout en minuscules
    return s.strip().lower().replace(' ', '_')

# 1) Chargement
train = pd.read_csv('dataset_project_iapr2025/train.csv')
labels = pd.read_csv('labels_YOLO_combined.csv')

# 2) Extraction id + type dans annotations
labels['id'] = labels['filename'].str.extract(r'(\d+)').astype(int)
labels['Choco_Type'] = labels['region_attributes'].apply(lambda x: json.loads(x)['Choco_Type'])

# 3) Comptage par image/type depuis les points
label_counts = labels.groupby(['id','Choco_Type']).size().unstack(fill_value=0)

# 4) Normalisation des noms de colonnes
#   a) sur label_counts : Jelly_White → jelly_white, etc.
label_counts_norm = label_counts.copy()
label_counts_norm.columns = [normalize_name(c) for c in label_counts_norm.columns]

#   b) sur train.csv : Crème brulée → creme_brulee, etc.
train_norm = train.rename(
    columns={c: normalize_name(c) for c in train.columns if c != 'id'}
).set_index('id')

# 5) Alignement des deux DataFrames
train_counts = (
    train_norm
    .reindex(label_counts_norm.index)                   # mêmes id
    .fillna(0)
    .astype(int)[label_counts_norm.columns]              # mêmes colonnes normalisées
)

# 6) Comparaison
diff = train_counts.compare(label_counts_norm)

if diff.empty:
    print("✅ Les deux sources sont totalement cohérentes.")
else:
    print("⚠️ Divergences détectées :")
    print(diff)
```

```python
from ast import literal_eval
import pandas as pd
import numpy as np
# Create class dict
name_list = ['Jelly_White','Jelly_Milk','Jelly_Black','Amandina','Creme_brulee','Triangolo','Tentation_noir','Comtesse','Noblesse','Noir_authentique','Passion_au_lait','Arabia','Stracciatella']
class_dict = {}
for i, name in enumerate(name_list) :
    class_dict[name] = i

class_dict
df = pd.read_csv('labels_YOLO_v2.csv')
df

# 2) Fonction de parsing sécurisée
def parse_shape(s):
    if isinstance(s, str):
        try:
            return literal_eval(s)
        except (ValueError, SyntaxError):
            # si jamais c'est du JSON plutôt que du dict Python
            import json
            return json.loads(s)
    elif isinstance(s, dict):
        return s
    else:
        # fallback vide
        return {}

def parse_attr(s):
    if isinstance(s, str):
        try:
            return literal_eval(s)
        except (ValueError, SyntaxError):
            import json
            return json.loads(s)
    elif isinstance(s, dict):
        return s
    else:
        return {}

# 3) Appliquer le parsing
df['region_shape_attributes'] = df['region_shape_attributes'].apply(parse_shape)
df['region_attributes']       = df['region_attributes'].apply(parse_attr)

# 4) Extraction centrée + dimensions
def extract_center_and_size(shape):
    # si c'est un point
    if 'cx' in shape and 'cy' in shape:
        cx, cy = shape['cx'], shape['cy']
    # sinon on calcule pour un rect
    else:
        cx = shape.get('x', 0) + shape.get('width', 0)/2
        cy = shape.get('y', 0) + shape.get('height', 0)/2
    return pd.Series({
        'cx': cx,
        'cy': cy,
        'width': shape.get('width', None),
        'height': shape.get('height', None)
    })

df = pd.concat([
    df,
    df['region_shape_attributes'].apply(extract_center_and_size)
], axis=1)

# 5) Classe numérique
df['class'] = df['region_attributes'].apply(lambda a: class_dict[a['Choco_Type']])

# Maintenant df a bien cx, cy, width, height et class sans ValueError

df
```

```python
cols_to_drop = [
    'file_size',
    'file_attributes',
    'region_count',
    'region_id',
    'region_shape_attributes',
    'region_attributes'
]
df = df.drop(columns=cols_to_drop, errors='ignore')

# Ensuite, supprimez toutes les colonnes en double (mêmes noms)
df = df.loc[:, ~df.columns.duplicated()]
df
```

```python
df.to_csv('labels_YOLO_v2.csv', index=False)
```

# Check images are ok :

```python
import os
import pandas as pd
from PIL import Image, ImageDraw

# 1) Chemins à adapter
csv_path     = 'labels_YOLO.csv'       # votre CSV avec filename,cx,cy,width,height,class
images_dir   = 'dataset_project_iapr2025/train'            # dossier contenant les JPG d'origine
output_dir   = 'Check_images'                        # dossier où seront sauvées les images annotées

# 2) Préparer le dossier de sortie (sans toucher aux originales)
os.makedirs(output_dir, exist_ok=True)

# 3) Charger le CSV
df = pd.read_csv(csv_path)

# 4) Parcourir chaque image
for filename, group in df.groupby('filename'):
    src_path = os.path.join(images_dir, filename)
    dst_path = os.path.join(output_dir,   filename)
    
    # Ouvrir l'image
    img = Image.open(src_path).convert('RGB')
    draw = ImageDraw.Draw(img)
    
    # Dessiner tous les rectangles de cette image
    for _, row in group.iterrows():
        cx, cy, w, h = row['cx'], row['cy'], row['width'], row['height']
        # coins du rectangle 
        x1 = int(cx - w/2)
        y1 = int(cy - h/2)
        x2 = int(cx + w/2)
        y2 = int(cy + h/2)
        # dessiner un contour rouge épaisseur 2
        draw.rectangle([x1, y1, x2, y2], outline='red', width=2)
    
    # Enregistrer l'image modifiée
    img.save(dst_path)
```
