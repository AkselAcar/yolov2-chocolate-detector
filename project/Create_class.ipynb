## Matching boxes

```python
# Import label_annotated
import pandas as pd
from ast import literal_eval
import numpy as np

df = pd.read_csv('label_annotated.csv')
df['region_shape_attributes'] = df['region_shape_attributes'].apply(literal_eval)
df['region_attributes'] = df['region_attributes'].apply(literal_eval)
df['rescaled_cx'] = df['region_shape_attributes'].apply(lambda x : int(np.rint(x['cx']/10)))
df['rescaled_cy'] = df['region_shape_attributes'].apply(lambda x : int(np.rint(x['cy']/10)))

df
```

```python
# Create class dict
name_list = ['Jelly_White','Jelly_Milk','Jelly_Black','Amandina','Creme_brulee','Triangolo','Tentation_noir','Comtesse','Noblesse','Noir_authentique','Passion_au_lait','Arabia','Stracciatella']
class_dict = {}
for i, name in enumerate(name_list) :
    class_dict[name] = i

class_dict
```

```python
# Convert region attributes into numbers
df['class'] = df['region_attributes'].apply(lambda x : class_dict[x['Choco_Type']])
df['width'] = df['region_attributes'].apply(lambda x : 50)
df['height'] = df['region_attributes'].apply(lambda x : 50)
df
```

```python
df.to_csv('labels_YOLO.csv', index=False)
```
