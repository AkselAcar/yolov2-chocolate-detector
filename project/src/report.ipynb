# Methodology

## Dataset:

The furnished dataset was composed of 13 references images (for the 13 classes), with one chocolate by image on a blanck background.
The train set was composed of 90 images of chocolates with different backgrounds, non-chocolate objects, and partially covered chocolates.
The test set was composed of 180 images of the same type as the train set. Also, every image is 6000x4000 pixels.

Moreover, the ground truth was given in a csv file, each row corresponding to an image of the train set, and the 13 last columns corresponding to the number of chocolates of each type in the associated image. However, we decided to use a YOLO algorithm, that requires bounding boxes for each chocolates, associated with the right label. Therefore, we used the tool VGG Image Annotator (https://www.robots.ox.ac.uk/~vgg/software/via/) to create the bounding boxes.

Then, we just had to reshape a little bit the created csv file, to keep only the interesting columns. There is one row by chocolate, first column being the name of the picture, and then cx and cy, the center coordinates, then the width and height of the bounding box. Finally the type of chocolate with an integer between 0 and 12. Moreover, everything is in pixel, with respect to the original images (6000x4000). We also had to shift cx and cy because by default, VGG image Annotator gives the upper-left corner of the box, while the convention for YOLO is the center. This labels can be found in the file `labels_YOLO_combined.csv`.

Finally, the file `Create_labels_and_link_to_images.py` has two goals. First, it creates a folder `labels` inside which there is one file by image of the training set. In each file there is one row by chocolate and every coordinate is normalized between 0 and 1 (center, width and height of each chocolate box). The second task is to create a link between the images names (.JPG) and their labels (.txt), this is more convenient for the rest of the code. It generates a csv file `link_images_labels_YOLO_v3_combined.csv` with two columns, each row correpsonding to one image name, one with a `.JPG`, the other with a `.txt`.

## YOLO: brief overview

YOLO (You Only Look Once) is a family of object detection algorithms designed for real-time applications. Unlike traditional methods that process candidate regions in multiple steps, YOLO reframes object detection as a single regression problem, directly predicting bounding boxes and class probabilities from the entire image in one pass.

YOLO uses a neural network in order to simultaneously predict different features of the objects to detect. Indeed, the neural networks takes an image as input, and by several convolutional layers and reductions, it outputs a grid of a predefined size. Usually, we try to get cells of the size of the objects.

In each cell, the model should predict : the confidence of having an object to detect inside, the center coordinates of the object with respect to the cell, the width and height of the bounding box, and a vector of the size of the number of classes, that should ideally be a one-hot encoding vector predicting the right class.



In our case, we implemented from scratch YOLO v2. It differs from the overview above because it uses what we call anchors. They are predefined boxes, and the model will try to align its prediction with them. To define them, we had to implement a K-Means algorithm on every bounding boxes of our training set.
